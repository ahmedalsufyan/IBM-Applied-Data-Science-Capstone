### Overview
Running structured SQL queries against SQLite database to compute launch site metrics and success rates.

```python
import sqlite3
import pandas as pd

# 1. Create and connect to an in-memory SQLite database
conn = sqlite3.connect('Spacex.db')
cursor = conn.cursor()

# 2. Load dataset and store it in a table named SPACEXTBL inside SQLite
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
df_sql = pd.read_csv(dataset_url)
df_sql.to_sql('SPACEXTBL', conn, if_exists='replace', index=False)

# 3. Example SQL queries for Exploratory Data Analysis (EDA)

# Query 1: Display distinct launch site names
query_1 = "SELECT DISTINCT LaunchSite FROM SPACEXTBL"
print("--- Distinct Launch Sites ---")
print(pd.read_sql_query(query_1, conn))
print("\n" + "="*40 + "\n")

# Query 2: Total payload mass carried by each launch site
query_2 = """
SELECT LaunchSite, SUM(PayloadMass) AS Total_Payload
FROM SPACEXTBL
GROUP BY LaunchSite
"""
print("--- Total Payload Mass per Launch Site ---")
print(pd.read_sql_query(query_2, conn))
print("\n" + "="*40 + "\n")

# Query 3: Landing success rates for each launch site
query_3 = """
SELECT LaunchSite,
       COUNT(*) AS Total_Launches,
       SUM(Class) AS Successful_Landings,
       ROUND(AVG(Class) * 100, 2) AS Success_Rate_Percentage
FROM SPACEXTBL
GROUP BY LaunchSite
"""
print("--- Success Rates per Launch Site ---")
print(pd.read_sql_query(query_3, conn))

# Close database connection
conn.close()
```
