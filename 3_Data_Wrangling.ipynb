### Overview
Cleaning dataset, imputing missing values, and engineering the binary target column `Class` (1 = Successful Landing, 0 = Unsuccessful Landing).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load dataset to ensure data is available in memory
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
df = pd.read_csv(dataset_url)

# 2. Handle missing values in PayloadMass if any exist
mean_payload = df['PayloadMass'].mean()
df['PayloadMass'] = df['PayloadMass'].fillna(mean_payload)

# 3. Ensure Class column is formatted as a binary integer (1 = Success, 0 = Failure)
df['Class'] = df['Class'].astype(int)

# 4. Generate the scatter plot
plt.figure(figsize=(12, 5))
sns.scatterplot(
    x="FlightNumber",
    y="PayloadMass",
    hue="Class",
    data=df,
    palette="Set1",
    s=100
)

plt.xlabel("Flight Number", fontsize=12)
plt.ylabel("Payload Mass (kg)", fontsize=12)
plt.title("Payload Mass vs Flight Number by Landing Outcome", fontsize=14)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```
