### Overview
Plotting payload relationships and flight trends using Seaborn and Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load dataset
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
df = pd.read_csv(dataset_url)

# Plot configuration setup
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 1, figsize=(12, 10))

# Plot 1: Flight Number vs Launch Site by Landing Outcome
sns.scatterplot(
    ax=axes[0],
    data=df,
    x="FlightNumber",
    y="LaunchSite",
    hue="Class",
    palette="Set1",
    s=100
)
axes[0].set_title("Flight Number vs. Launch Site by Landing Outcome", fontsize=14)
axes[0].set_xlabel("Flight Number", fontsize=12)
axes[0].set_ylabel("Launch Site", fontsize=12)

# Plot 2: Orbit Type vs Landing Success Rate
orbit_success = df.groupby('Orbit')['Class'].mean().reset_index()
sns.barplot(
    ax=axes[1],
    data=orbit_success,
    x="Orbit",
    y="Class",
    palette="Blues_d"
)
axes[1].set_title("Success Rate by Orbit Type", fontsize=14)
axes[1].set_xlabel("Orbit Type", fontsize=12)
axes[1].set_ylabel("Success Rate (0.0 to 1.0)", fontsize=12)

plt.tight_layout()
plt.show()
```
