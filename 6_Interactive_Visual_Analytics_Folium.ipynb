### Overview
Visualizing geographical positions of SpaceX launch sites and clustering landing outcomes.

```python
import folium
import pandas as pd
from folium.plugins import MarkerCluster

# 1. Load dataset containing geographical coordinates (Latitude / Longitude)
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"
df_map = pd.read_csv(dataset_url)

# 2. Create an interactive map centered over the United States
site_map = folium.Map(location=[30.6, -81.5], zoom_start=4)

# 3. Create a MarkerCluster object to group landing outcomes
marker_cluster = MarkerCluster().add_to(site_map)

# 4. Add each launch record to the map (Green = Success, Red = Failed)
for index, row in df_map.iterrows():
    # Set marker color based on Class (1 = Success, 0 = Failed)
    marker_color = 'green' if row['Class'] == 1 else 'red'
    popup_text = f"Site: {row['LaunchSite']}<br>Outcome: {'Success' if row['Class'] == 1 else 'Failed'}"

    # Add marker to cluster
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        icon=folium.Icon(color='white', icon_color=marker_color),
        popup=popup_text
    ).add_to(marker_cluster)

# Display the interactive map
site_map
```
